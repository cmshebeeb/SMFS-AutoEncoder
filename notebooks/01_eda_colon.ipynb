#### Plot 4 random feature histograms

```python
import random

random_features = random.sample(list(df.columns[:-1]), 4)

df[random_features].hist(figsize=(10, 6))
plt.tight_layout()
plt.show()
```

All four features (feature_506, feature_494, feature_1351, and feature_1814) exhibit a similar three-cluster distribution with peaks around -2, 0, and 2, indicating discretized or standardized feature values. \
The negative cluster (-2) has the highest frequency across most features, while the positive cluster (2) consistently has the lowest frequency. \
The similar distribution patterns suggest these features may be highly correlated or derived from the same preprocessing/transformation step.
